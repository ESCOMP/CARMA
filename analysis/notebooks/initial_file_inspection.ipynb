```python
import xarray as xr
import matplotlib.pyplot as plt
import netCDF4
import scipy
import cftime
```

```python
# SW and LW range for plot
l_sw_min = 0.3e-4 # cm
l_sw_max = 3.0e-4 # cm
l_lw_min = 3.0e-4 # cm
l_lw_max = 100.0e-4 # cm

def add_SW_LW_plot_areas(ax):
    ax.axvspan(l_sw_min, l_sw_max, facecolor='lightskyblue', alpha=0.2, label='Shortwave region')
    ax.axvspan(l_lw_min, l_lw_max , facecolor='lightcoral', alpha=0.2, label='Longwave region')

def wavelength_x(ax):
    ax.set_xlabel(f'Wavelength $\lambda$ / cm')

    
```

### Alumina optical properties - alumina.nc 

```python
ds = xr.open_dataset('/home/n02/n02/hqmp0145/CARMA/analysis/inputdata/Alumina.nc', decode_times=False)
```

```python
fig, ax = plt.subplots()

ds['real'].plot(
    ax=ax,
    marker='.',
    linestyle='None',
    xscale='log'
)

add_SW_LW_plot_areas(ax)
plt.legend()

wavelength_x(ax)
ax.set_ylabel('Real refractive index')
ax.set_title('Alumina refractive index')
ax.grid()
plt.show()
```
