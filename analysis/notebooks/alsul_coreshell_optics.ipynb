```python
import xarray as xr
import matplotlib.pyplot as plt
import functions as fns
```

```python
ds = xr.open_dataset('/home/n02/n02/hqmp0145/CARMA/run/carma/carma_alsul_coreshell_optics_test_cf90,99,1.nc', decode_times=False)
```

```python
ds = ds.sortby("wave")
```

```python
ds
```

```python
fig, ax = plt.subplots(2, 1, sharex=True, figsize=(8, 8))

variable_label = {'PRSUL': 'Pure sulphate', 
                  'ALSUL_fixed_wet': 'Al-Sul wet - fixed core', 
                  'ALSUL_fixed_dry': 'Al-Sul dry - fixed core',
                  'ALSUL_coreshell_cf90_wet': 'Al-Sul wet - core/shell cf90', 
                  'ALSUL_coreshell_cf90_dry': 'Al-Sul dry - core/shell cf90',
                  'ALSUL_coreshell_cf99_wet': 'Al-Sul wet - core/shell cf99', 
                  'ALSUL_coreshell_cf99_dry': 'Al-Sul dry - core/shell cf99',
                  'ALSUL_coreshell_cf1_wet': 'Al-Sul wet - core/shell cf1', 
                  'ALSUL_coreshell_cf1_dry': 'Al-Sul dry - core/shell cf1',
                  }

variables = [
    'PRSUL', 
             'ALSUL_fixed_wet', 
             'ALSUL_fixed_dry', 
             'ALSUL_coreshell_cf90_wet', 
             'ALSUL_coreshell_cf90_dry',
            'ALSUL_coreshell_cf99_wet', 
            'ALSUL_coreshell_cf99_dry',
            # 'ALSUL_coreshell_cf**_wet',
            # 'ALSUL_coreshell_cf**_dry',
             ]

for variable in variables:
# Top: imaginary refractive index
    ds[f'{variable}_refidx_i'].plot(
        ax=ax[0],
        xscale='log',
        # alpha=0.6,
        # ls='--',
        label=variable_label[variable]
    )

    # Bottom: real refractive index
    ds[f'{variable}_refidx_r'].plot(
        ax=ax[1],
        xscale='log'
    )

# Wavelength axis / SW-LW regions
fns.wavelength_x(ax[1])
ax[0].set_xlabel('')
fns.add_SW_LW_plot_areas(ax[0])
fns.add_SW_LW_plot_areas(ax[1])

ax[0].legend()

ax[0].grid()
ax[1].grid()
ax[0].set_ylabel("Imaginary refractive index, k")
ax[1].set_ylabel("Real refractive index, n")

plt.tight_layout()
```
